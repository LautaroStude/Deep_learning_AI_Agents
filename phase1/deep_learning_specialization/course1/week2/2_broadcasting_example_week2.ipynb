```python
A = np.array([[56.0, 0.0, 4.4, 68.0],
              [1.2, 104.0, 52.0, 8.0],
              [1.8, 135.0, 99.0, 0.9]])
print(A)
```

```python
sum_col = A.sum(axis=0)
print(sum_col)
```

```python
rehacer_matriz = sum_col.reshape(1,4)
print(rehacer_matriz)
```

```python
percentage = 100*A/rehacer_matriz
print(percentage)

#divide el numero de la matriz (0,0) con cada fila de la misma columna 0 en A, divide 56 por 59, 1,2 por 59 y 1,8 con 59 asi pasa
#a la otra columna
```
